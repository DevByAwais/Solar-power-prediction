```python
#Feature Scaling
def scale_features(X_train, X_test):

    mean = X_train.mean()
    std = X_train.std()

    X_train_scaled = (X_train - mean) / std
    X_test_scaled = (X_test - mean) / std

    return X_train_scaled, X_test_scaled, mean, std
```

```python
#Scale Set A
X_train_A_scaled, X_test_A_scaled, mean_A, std_A = scale_features(
    X_train_A,
    X_test_A
)

print("Set A training mean:")
print(mean_A)

print("\nSet A training standard deviation:")
print(std_A)
```

```python
#Part 6  Define the required functions
#Linear Regression Functions from Scratch

import numpy as np

# Hypothesis function
def hypothesis(X, theta):
    return X @ theta


# Cost function
def cost(X, y, theta):
    predictions = hypothesis(X, theta)
    errors = predictions - y
    J = 0.5 * np.sum(errors ** 2)
    return J


# Normal Equation
def fit_normal(X, y):
    theta = np.linalg.inv(X.T @ X) @ X.T @ y
    return theta


# Batch Gradient Descent
def fit_batch_gd(X, y, alpha, iterations):
    theta = np.zeros(X.shape[1])
    cost_history = []

    for i in range(iterations):
        predictions = hypothesis(X, theta)
        errors = predictions - y

        gradient = X.T @ errors

        theta = theta - alpha * gradient

        J = cost(X, y, theta)
        cost_history.append(J)

    return theta, cost_history


# Stochastic Gradient Descent
def fit_sgd(X, y, alpha, epochs):
    theta = np.zeros(X.shape[1])
    cost_history = []

    for epoch in range(epochs):

        for i in range(len(y)):
            xi = X[i]
            yi = y[i]

            prediction = xi @ theta
            error = prediction - yi

            theta = theta - alpha * xi * error

        J = cost(X, y, theta)
        cost_history.append(J)

    return theta, cost_history


# RMSE function
def rmse(y_actual, y_predicted):
    error = y_predicted - y_actual
    return np.sqrt(np.mean(error ** 2))


print("All regression functions are ready.")
```

```python
#Part 7  Test Normal Equation on Set A
# Part 7: Normal Equation - Set A

theta_A_normal = fit_normal(X_train_A_final, y_train)

print("Theta for Set A using Normal Equation:")
print(theta_A_normal)

# Training prediction
train_pred_A_normal = hypothesis(X_train_A_final, theta_A_normal)

# Test prediction
test_pred_A_normal = hypothesis(X_test_A_final, theta_A_normal)

# Clip negative predictions to zero
train_pred_A_normal = np.maximum(train_pred_A_normal, 0)
test_pred_A_normal = np.maximum(test_pred_A_normal, 0)

# RMSE
train_rmse_A_normal = rmse(y_train, train_pred_A_normal)
test_rmse_A_normal = rmse(y_test, test_pred_A_normal)

print()
print("Training RMSE:", train_rmse_A_normal)
print("Test RMSE:", test_rmse_A_normal)
```

```python
#Part 8  Batch GD learning-rate experiment

learning_rates_batch = [1e-5, 1e-4, 1e-3]

batch_results = {}

for alpha in learning_rates_batch:

    theta, cost_history = fit_batch_gd(
        X_train_A_final,
        y_train,
        alpha,
        500
    )

    batch_results[alpha] = {
        "theta": theta,
        "cost_history": cost_history
    }

    print("Learning Rate:", alpha)
    print("Initial Cost:", cost_history[0])
    print("Final Cost:", cost_history[-1])
    print()
```

```python
#Part 33  Create the Prediction Function

theta = np.load("model/theta_setB.npy")
feature_mean = np.load("model/setB_mean.npy")
feature_std = np.load("model/setB_std.npy")


def predict_ac_power(hour, sw_radiation, temp_2m, cloud_cover):

    # Create weather features
    features = np.array([
        sw_radiation,
        temp_2m,
        cloud_cover
    ])

    # Scale using training statistics
    features_scaled = (
        features - feature_mean[:3]
    ) / feature_std[:3]

    # Time features
    sin_hour = np.sin(2 * np.pi * hour / 24)
    cos_hour = np.cos(2 * np.pi * hour / 24)

    # Create input with intercept
    X = np.array([
        1,
        features_scaled[0],
        features_scaled[1],
        features_scaled[2],
        sin_hour,
        cos_hour
    ])

    # Prediction
    prediction = X @ theta

    # Clip negative prediction
    prediction = max(prediction, 0)

    return prediction
```
